```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

plt.figure(figsize=(6,4))
sns.histplot(df['sepal length (cm)'], bins=20, kde=True)
plt.title('Distribution of Sepal Length')
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(x = df['sepal width (cm)'])
plt.title('boxplot of Sepal length')
plt.show()

speices_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(speices_counts, labels=iris.target_names, autopct='%1.1f%%')
plt.title('Species Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'])
plt.title('Sepal Length vs Sepal Width')
plt.show()

plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title('Sepal Length Over Samples')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()

plt.figure(figsize=(6,4))
sns.boxplot(x=df['species'], y=df['petal length (cm)'])
plt.title('Boxplot of Petal Length by Species')
plt.xticks([0, 1, 2], iris.target_names)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Iris Features')
plt.show()

plt.figure(figsize=(6,4))
plt.scatter(df['petal length (cm)'], df['petal width (cm)'], c=df['species'])
plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()  

sns.heatmap(df.iloc[:,:4].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Iris Features')
plt.show()  

plt.figure(figsize=(6,4))
plt.scatter(
    df['sepal length (cm)'],
    df['petal length (cm)'],
    s=df['sepal width (cm)'] * 20,
    alpha=0.5,
    c=df['species']
)
plt.title('Sepal Length vs Petal Length with Sepal Width as Size')
plt.xlabel('Sepal Length (cm)') 
plt.ylabel('Petal Length (cm)')
plt.show()

sns.pairplot(df.iloc[:,:4])
plt.suptitle('Pairplot of Iris Features', y=1.02)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Violin Plot of Sepal Length by Species')
plt.xticks([0, 1, 2], iris.target_names)    
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load tips.csv
# Load the built-in Seaborn tips dataset
df = sns.load_dataset('tips')

# Display first few rows
print(df.head())

# 1. Distribution of Total Bill
plt.figure(figsize=(6,4))
sns.histplot(df['total_bill'], bins=20, kde=True)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()

# 2. Distribution of Tip
plt.figure(figsize=(6,4))
sns.histplot(df['tip'], bins=20, kde=True)
plt.title('Distribution of Tip')
plt.xlabel('Tip')
plt.ylabel('Frequency')
plt.show()

# 3. Day-wise distribution
day_counts = df['day'].value_counts()

plt.figure(figsize=(6,6))
plt.pie(day_counts, labels=day_counts.index, autopct='%1.1f%%')
plt.title('Day Distribution')
plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.scatterplot(x=df['total_bill'], y=df['tip'], hue=df['sex'])

plt.title('Total Bill vs Tip')

plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()


plt.figure(figsize=(6,4))

plt.plot(df['total_bill'])

plt.title('Total Bill Over Samples')

plt.xlabel('Sample Index')
plt.ylabel('Total Bill')

plt.show()


plt.figure(figsize=(6,4))

sns.boxplot(x=df['day'], y=df['tip'])

plt.title('Boxplot of Tip by Day')

plt.xlabel('Day')
plt.ylabel('Tip')

plt.show()
```
